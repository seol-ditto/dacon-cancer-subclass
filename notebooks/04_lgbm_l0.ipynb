```python
import sys; sys.path.append('..')

import lightgbm as lgb

from src.config import SEED
from src.data import load_data, get_gene_cols
from src.features import to_binary
from src.validation import seed_everything, encode_target, run_cv

seed_everything()
train, test, submission = load_data()
genes = get_gene_cols(train)
y, le = encode_target(train)

X = to_binary(train, genes).astype('float32')
print(X.shape)
```

```python
params = dict(
    objective='multiclass', n_estimators=300, learning_rate=0.05,
    num_leaves=31, min_child_samples=10,
    colsample_bytree=0.3, subsample=0.8, subsample_freq=1,
    class_weight='balanced', random_state=SEED, n_jobs=-1, verbose=-1,
)
result = run_cv(lambda: lgb.LGBMClassifier(**params), X, y)   # 비교용이라 X_test 생략
```
