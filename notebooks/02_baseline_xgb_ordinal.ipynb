```python
import sys; sys.path.append('..')

import xgboost as xgb
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OrdinalEncoder

from src.config import SEED
from src.data import load_data, get_gene_cols
from src.validation import seed_everything, encode_target, run_cv

seed_everything()
train, test, submission = load_data()
genes = get_gene_cols(train)
y, le = encode_target(train)

# 제공 베이스라인과 같은 설정. OrdinalEncoder는 Pipeline으로 fold 안에서만 fit
make_model = lambda: make_pipeline(
    OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1),
    xgb.XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=6,
                      random_state=SEED, eval_metric='mlogloss', n_jobs=-1),
)
result = run_cv(make_model, train[genes], y)   # 참고용이라 X_test 생략 (제출 안 함)
```
